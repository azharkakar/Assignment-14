```python
#Question 5: Draw a bar chart showing average order value by city.
#Answer:
avg_city = df.groupby('city')['order_value'].mean()
```

```python
import matplotlib.pyplot as plt
categories = avg_city.index
values = avg_city.values
plt.bar(categories, values, color='pink', width=0.5)
plt.xlabel('City')
plt.ylabel('Average Order Value')
plt.title('Average Order Value by City')
plt.show()
```
